```python
#matrix multipilication or Dot product
tensor1 = torch.rand(3, 4)
tensor2 = torch.rand(4, 2)
tensor1, tensor2
```

```python
#torch.matmul
%%time
tensorM = torch.matmul(tensor1, tensor2)
tensorM
```

```python
#loops
%%time
result = torch.zeros(size=(len(tensor1), len(tensor2[0])))
for i in range(len(tensor1)):
  for j in range(len(tensor2[0])):
    value = 0
    for k in range(len(tensor2)):
      value += tensor1[i][k] * tensor2[k][j]
    result[i][j] = value

result
```

```python
# @ symbol
%%time
tensor1 @ tensor2
```

```python
# Shapes for matrix multiplication
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]])

tensor_B = torch.tensor([[7, 10],
                         [8, 11],
                         [9, 12]])
```

```python
#transpose
tensor_B.T, tensor_B.T.shape
```

```python
torch.mm(tensor_A, tensor_B.T)
```

```python
print(f"tensor_A.shape: {tensor_A.shape}, tensor_B.shape: {tensor_B.shape}")
print(f"New Shape: tensor_A.shape: {tensor_A.shape}, tensor_B.T.shape: {tensor_B.T.shape}")
print(f"Multiplying: {tensor_A.shape} @ {tensor_B.T.shape} <- inner dimensions must match")
print("Output:\n")
output = torch.mm(tensor_A, tensor_B.T)
print(output)
print(f"\nOutput shape: {output.shape}")
```

## Reshaping, stacking, squeezing and unsqueezing tensors


*   Reshaping - reshaping an input tensor to a defined shape
*   View - Return a view of an input tensor of a certain shape but keep the same memory as the original tensor
*   Stacking - combine multiple tensors on top of each other (vstack) or side by side (hstack)
*   Squeeze - removes all '1' dimensions from a tensor
*   Unsqueeze - add a '1' dimension to a target tensor
*   Permute - Return a view of the input with dimensions permuted (swapped) in a certain way



```python
import torch
x = torch.arange(1., 10.)
x, x.shape
```

```python
#Add an extra dimension
x_reshaped = x.reshape(1, 9)
x_reshaped, x_reshaped.shape
```

```python
# Change the view-- changing z changes x because view of a tensor shares the same memory as the original input)
z = x.view(1, 9)
z, z.shape
```

```python
z[:,0] = 5
z, x
```

```python
# Stack tensors on top of x
torch.vstack([x, x, x, x]), torch.hstack([x, x, x, x])
```

```python
torch.stack([x, x, x, x], dim = 0), torch.stack([x, x, x, x], dim = 1)
```

## Reproducibility (trying to take random out of random)

In short how a neural network learns:

`start with random numbers -> tensor operations -> update random numbers to try and make better representations of the data -> REPEAT...`

To reduce the randomness in nn and pytorch comes the concept of a **random seed**.

Essentially what the random seed does is "flavour" the randomness.

```python
import torch

# create two random tensors
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python
# Random but reproducible tensor

# Set the random seed
RANDOM_SEED  = 42

torch.manual_seed(RANDOM_SEED)
random_tensor_C = torch.rand(3, 4)

torch.manual_seed(RANDOM_SEED)
random_tensor_D = torch.rand(3, 4)

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)
```
